# This is a notebook where error anlaysis will be performed on our best model

```python
import pandas as pd
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix,
    classification_report, average_precision_score,
    matthews_corrcoef, cohen_kappa_score
)
```

```python
engine_wiring_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_1\engine_wiring_error_analysis.csv')
pipe_clip_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_1\pipe_clip_error_analysis.csv')
pipe_staple_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_2\pipe_staple_error_analysis.csv')
tank_screw_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_2\tank_screw_error_analysis.csv')
underbody_pipes_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_3\underbody_pipes_error_analysis.csv')
underbody_screw_csv = pd.read_csv(r'C:\Users\victo\Desktop\FE_Paper\automotive-anomaly-detection\results\results-2026.03.15-federated-kcenter-new-interpretability\clients\client_3\underbody_screw_error_analysis.csv')
```

```python
df = pd.concat([engine_wiring_csv,pipe_clip_csv,pipe_staple_csv,tank_screw_csv,underbody_pipes_csv,underbody_screw_csv])
y_true   = df["true_label"]
y_pred   = df["predicted_label"]
y_scores = df["anomaly_score"]   # continuous score for AUC curves

# ── Core metrics ───────────────────────────────────────────
acc       = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, zero_division=0)
recall    = recall_score(y_true, y_pred, zero_division=0)
f1        = f1_score(y_true, y_pred, zero_division=0)
f1_macro  = f1_score(y_true, y_pred, average="macro",    zero_division=0)
f1_w      = f1_score(y_true, y_pred, average="weighted", zero_division=0)

# ── Score-based metrics (use anomaly_score) ────────────────
roc_auc   = roc_auc_score(y_true, y_scores)
pr_auc    = average_precision_score(y_true, y_scores)  # area under PR curve

# ── Extra metrics useful for imbalanced data ───────────────
mcc       = matthews_corrcoef(y_true, y_pred)
kappa     = cohen_kappa_score(y_true, y_pred)

# ── Confusion matrix ───────────────────────────────────────
tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
specificity = tn / (tn + fp)

# ── Print results ──────────────────────────────────────────
print("=" * 45)
print(f"  Accuracy          : {acc:.4f}")
print(f"  Precision         : {precision:.4f}")
print(f"  Recall (Sensitivity): {recall:.4f}")
print(f"  Specificity       : {specificity:.4f}")
print(f"  F1 (binary)       : {f1:.4f}")
print(f"  F1 (macro)        : {f1_macro:.4f}")
print(f"  F1 (weighted)     : {f1_w:.4f}")
print(f"  ROC-AUC           : {roc_auc:.4f}")
print(f"  PR-AUC            : {pr_auc:.4f}")
print(f"  MCC               : {mcc:.4f}")
print(f"  Cohen's Kappa     : {kappa:.4f}")
print("=" * 45)
print(f"\n  Confusion Matrix  : TP={tp} FP={fp} FN={fn} TN={tn}")
print("\nClassification Report:\n")
print(classification_report(y_true, y_pred, zero_division=0))
```
